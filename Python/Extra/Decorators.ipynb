# Decorators in Python

Decorators are one of Python's most powerful features. They allow you to **modify or extend the behavior of a function or class without changing its source code**.

Decorators are widely used in Python libraries such as **TensorFlow**, **Flask**, **FastAPI**, **Django**, and **pytest**.

---

# What is a Decorator?

A **decorator** is simply a function that takes another function as input, adds some functionality, and returns a new function.

Conceptually:

```text
Original Function
        │
        ▼
   Decorator
        │
        ▼
Modified Function
```

Instead of modifying the original function directly, the decorator **wraps** it with additional behavior.

---

# Why are Decorators Useful?

Decorators allow you to reuse common functionality without rewriting code.

Typical use cases include:

- Logging function calls
- Measuring execution time
- Authentication and authorization
- Input validation
- Caching results
- Registering routes in web frameworks
- Optimizing TensorFlow functions

Rather than modifying every function individually, you write the additional behavior once inside a decorator.

---

# Functions are First-Class Objects

In Python, functions are **first-class objects**, which means they can:

- Be assigned to variables.
- Be passed as arguments.
- Be returned from other functions.

This is what makes decorators possible.

---

# Creating a Basic Decorator

A decorator is simply another function that accepts a function and returns a new one.

```python
def decorator(func):

    def wrapper():

        print("Before calling the function")

        func()

        print("After calling the function")

    return wrapper


@decorator
def greet():
    print("Hello!")


greet()
```

Output:

```text
Before calling the function
Hello!
After calling the function
```

The line:

```python
@decorator
```

is equivalent to writing:

```python
greet = decorator(greet)
```

---

# Decorators with Function Arguments

Most functions receive parameters.

To make decorators work with any function signature, use `*args` and `**kwargs`.

```python
def decorator(func):

    def wrapper(*args, **kwargs):

        print("Before")

        result = func(*args, **kwargs)

        print("After")

        return result

    return wrapper


@decorator
def add(a, b):
    return a + b


print(add(5, 3))
```

Output:

```text
Before
After
8
```

- `*args` collects positional arguments.
- `**kwargs` collects keyword arguments.

This allows the decorator to wrap almost any function.

# Example: Measuring Execution Time

Decorators are commonly used for timing functions.

```python
import time

def timer(func):

    def wrapper(*args, **kwargs):

        start = time.time()

        result = func(*args, **kwargs)

        end = time.time()

        print(f"Execution time: {end - start:.4f} seconds")

        return result

    return wrapper


@timer
def slow_function():

    time.sleep(2)

    print("Finished!")


slow_function()
```

Output:

```text
Finished!
Execution time: 2.0015 seconds
```

# Decorators in TensorFlow

One of the most common decorators in TensorFlow is:

```python
@tf.function
```

Example:

```python
import tensorflow as tf

@tf.function
def square(x):
    return x * x
```

Without `@tf.function`, TensorFlow executes operations immediately using **Eager Execution**.

With `@tf.function`, TensorFlow:

1. Records the operations.
2. Converts them into a computational graph.
3. Optimizes the graph.
4. Executes the graph more efficiently.

Conceptually:

```text
Python Function
       │
       ▼
@tf.function
       │
       ▼
TensorFlow Graph
       │
       ▼
Optimized Execution
```

This optimization is especially beneficial when training large neural networks or performing repeated computations.

# Preserving Function Metadata

Decorators replace one function with another, which can hide the original function's name and documentation.

Python provides the `functools.wraps` decorator to preserve this metadata.

```python
from functools import wraps

def decorator(func):

    @wraps(func)
    def wrapper(*args, **kwargs):

        print("Before")

        result = func(*args, **kwargs)

        print("After")

        return result

    return wrapper
```

Using `@wraps` preserves attributes such as:

- Function name (`__name__`)
- Documentation (`__doc__`)
- Module information

# Common Python Decorators

| Decorator | Purpose |
|-----------|---------|
| `@property` | Access a method like an attribute |
| `@staticmethod` | Define a method that doesn't use `self` |
| `@classmethod` | Define a method that receives the class (`cls`) |
| `@dataclass` | Automatically generate methods like `__init__()` |
| `@lru_cache` | Cache function results |
| `@tf.function` | Convert TensorFlow code into an optimized computation graph |

# Summary

A **decorator** is a function that takes another function as input, adds additional behavior, and returns a new function.

Decorators make code more modular, reusable, and maintainable by separating cross-cutting concerns—such as logging, timing, caching, or optimization—from the core logic of a function.

In TensorFlow, decorators play an important role. The `@tf.function` decorator transforms ordinary Python functions into optimized computational graphs, allowing TensorFlow to execute machine learning models more efficiently while keeping the code clean and easy to read.

````python
class Car:
    number_of_cars = 0

    def __init__(self, brand, speed):
        self.brand = brand
        self._speed = speed #protected attribute 
        Car.number_of_cars += 1

    @property
    def speed(self):
        return self._speed

    @speed.setter
    def speed(self, value):
        if value < 0:
            raise ValueError("Speed cannot be negative")
        self._speed = value

    def accelerate (self, amount):
        self._speed += amount

    @classmethod
    def total_cars(cls):
        return cls.number_of_cars
    
    @staticmethod
    def kmh_to_mph(speed):
        """Convert km/h to mph."""
        return speed * 0.621371
    
````

````python
car1 = Car("Volvo", 80)

# Instance method
car1.accelerate(20)
print(f"{car1.brand} speed: {car1.speed} km/h")

# @property setter
car1.speed = 150
print(f"{car1.brand} speed: {car1.speed} km/h")
# without property: car.set_speed(120)

# @staticmethod
mph = Car.kmh_to_mph(car1.speed)
print(f"{car1.speed} km/h = {mph:.2f} mph")

# @classmethod
print(f"Total cars created: {Car.total_cars()}")
````
